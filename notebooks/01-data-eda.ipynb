```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("/kaggle/input/datasets/bhanujachanda/patient-data/patient_records.csv")
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
Biomarker_Level_median = data["Biomarker_Level"].median()

data["Biomarker_Level"] = data["Biomarker_Level"].fillna(Biomarker_Level_median)

print(Biomarker_Level_median)
```

```python
data.isnull().sum()
```

```python
Prior_Reaction_History_mode = data["Prior_Reaction_History"].mode()[0]
data["Prior_Reaction_History"] = data["Prior_Reaction_History"].fillna(Prior_Reaction_History_mode)
```

```python
data.isnull().sum()
```

```python
data['Reaction_Score'].value_counts().plot(kind='bar')
plt.title('Reaction Score Distribution')
plt.xlabel('Reaction_score')
plt.ylabel('Number of patients')
plt.show()
```

```python
plt.hist(data['Age'],bins=100)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of patients')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=data, x="Vaccine_Dose_mcg",
                y="Biomarker_Level", hue="Reaction_Score")
plt.title("Vaccine Dose vs Biomarker Level")
plt.savefig(f"{"/kaggle/working/"}vaccine_dose_vs_biomarker.png",
            dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=data, x="Reaction_Score", y="Age")
plt.title("Age Distribution by Reaction Score")
plt.savefig(F"{"/kaggle/working/"}age_vs_reaction.png",
            dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=data, x="Reaction_Score", y="Biomarker_Level")
plt.title("Biomarker Level vs Reaction Score")
plt.savefig(f"{"/kaggle/working/"}biomarker_vs_reaction.png",
            dpi=300, bbox_inches="tight")
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=data, x="Reaction_Score")
plt.title("Reaction Score Distribution")
plt.xlabel("Reaction Score")
plt.ylabel("Number of Patients")
plt.savefig(f"{"/kaggle/working/"}/reaction_score_distribution.png",
            dpi=300, bbox_inches="tight")
plt.show()
```
